## Part C: Storage Technologies

Here different storage technologies will be added to the system. First a battery storage of different durations are tested. A 4, 6, and 8 hour battery. A 4 hour battery of 100 MW will have a capacity of 400 MWh.


```python
# Battery storage parameters

energy_cost = 110 # €/kWh
battery_hours = 4 # hours (change this to either 4, 6, or 8)
# PyPSA requires capital cost in €/kW, so we need to convert the energy cost to power cost
CAPEX_per_kW = energy_cost * battery_hours # €/kW

# This should be added to the PyPSA network
n.add("StorageUnit",
            name="battery",
            carrier="battery",
            bus="NLD",
            p_nom_extendable=True,
            max_hours=battery_hours,
            capital_cost=CAPEX_per_kW,
            marginal_cost=0,
            efficiency_store=0.95,
            efficiency_dispatch=0.95)
```

```python
# Hydrogen storage parameters

n.add("Carrier", "H2")

n.add("Bus",
    "hydrogen_bus",
    carrier="H2")

# Electrolyzer parameters
n.add("Link",
    "electrolyzer",
    bus0="electricity_bus",
    bus1="hydrogen_bus",
    efficiency=0.67,
    p_nom_extendable=True,
    capital_cost=548,   # €/kW
    carrier="H2 electrolysis")

# Hydrogen storage parameters
n.add("Store",
    "hydrogen_storage",
    bus="hydrogen_bus",
    e_nom_extendable=True,
    capital_cost=1,   # €/kWh
    standing_loss=0,
    carrier="H2 storage")

# Fuel cell parameters
n.add("Link",
    "fuel_cell",
    bus0="hydrogen_bus",
    bus1="electricity_bus",
    efficiency=0.55,
    p_nom_extendable=True,
    capital_cost=2610,   # €/kW
    carrier="H2 fuel cell")
```
